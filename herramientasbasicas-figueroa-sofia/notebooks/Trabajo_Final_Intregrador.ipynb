## 5. Limpieza y transformación de los datos

Se realizó una revisión de los tipos de datos, valores nulos y registros duplicados. No se encontraron valores nulos ni duplicados.

La columna Order Date se convirtió al formato de fecha para permitir el análisis temporal.

También se verificó la consistencia de Total Price mediante el cálculo de Quantity × Unit Price.

```
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Calculated Total"] = (
    df["Quantity"] * df["Unit Price"]
)

df["Difference"] = abs(
    df["Total Price"] - df["Calculated Total"]
)

print("Diferencia máxima:", df["Difference"].max())

print(
    "Registros con diferencia mayor a 0.01:",
    (df["Difference"] > 0.01).sum()
)
```

# Análisis de Outliers

Para identificar posibles valores atípicos se analizaron las variables numéricas Quantity, Unit Price y Total Price mediante el método del rango intercuartílico (IQR).

```
# Seleccionamos las columnas numéricas
numeric_columns = ["Quantity", "Unit Price", "Total Price"]

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"\nVariable: {column}")
    print("Cantidad de outliers:", len(outliers))
```

### Interpretación de outliers

Se analizaron las variables Quantity, Unit Price y Total Price mediante el método del rango intercuartílico (IQR). Los valores identificados como atípicos fueron revisados para determinar si representaban errores en los datos o valores reales dentro del comportamiento de las ventas.

Debido a que el dataset presenta valores coherentes y no se detectaron inconsistencias en el cálculo del precio total, los valores atípicos encontrados no fueron eliminados automáticamente.

# Visualizaciones y Análisis Exploratorio

## 1. Evolución mensual de las ventas

```
df["Order Date"] = pd.to_datetime(df["Order Date"])
# Crear columna de mes
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)

# Agrupar ventas por mes
monthly_sales = df.groupby("Month")["Total Price"].sum().reset_index()

plt.figure(figsize=(10, 5))

sns.barplot(
    data=monthly_sales,
    x="Month",
    y="Total Price"
)

plt.title("Evolución mensual de las ventas")
plt.xlabel("Mes")
plt.ylabel("Ventas totales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

### Interpretación

Las ventas se distribuyen de manera relativamente estable durante el año 2023. Sin embargo, algunos meses presentan valores superiores a otros. El mes de diciembre registró el mayor nivel de ventas, mientras que noviembre presentó uno de los niveles más bajos.

Este análisis permite identificar posibles variaciones mensuales en el comportamiento de las ventas.

## 2. Ventas totales por categoría de producto

```
category_sales = (
    df.groupby("Product Category")["Total Price"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=category_sales,
    x="Total Price",
    y="Product Category"
)

plt.title("Ventas totales por categoría")
plt.xlabel("Ventas")
plt.ylabel("Categoría")

plt.tight_layout()

plt.show()
```

### Interpretación

La categoría Sports & Outdoors registró el mayor volumen de ventas, seguida por Clothing y Electronics. Por otro lado, Beauty & Health presentó el menor volumen de ventas dentro de las categorías analizadas.

Las diferencias entre categorías permiten identificar cuáles son las áreas que generan una mayor contribución a las ventas totales del e-commerce.

## 3. Cantidad de pedidos por medio de pago

```
payment_orders = (
    df["Payment Type"]
    .value_counts()
    .reset_index()
)

payment_orders.columns = ["Payment Type", "Orders"]

plt.figure(figsize=(8, 5))

sns.barplot(
    data=payment_orders,
    x="Orders",
    y="Payment Type"
)

plt.title("Cantidad de pedidos por medio de pago")
plt.xlabel("Cantidad de pedidos")
plt.ylabel("Medio de pago")

plt.tight_layout()

plt.show()
```

### Interpretación

Los pedidos se encuentran distribuidos de manera bastante equilibrada entre los diferentes medios de pago. PayPal registró la mayor cantidad de pedidos, seguido por Debit Card.

Esta distribución indica que no existe una dependencia excesiva de un único medio de pago.

## 4. Distribución de los pedidos según su estado

```
status_orders = (
    df["Order Status"]
    .value_counts()
    .reset_index()
)

status_orders.columns = ["Order Status", "Orders"]

plt.figure(figsize=(8, 5))

sns.barplot(
    data=status_orders,
    x="Order Status",
    y="Orders"
)

plt.title("Cantidad de pedidos según su estado")
plt.xlabel("Estado del pedido")
plt.ylabel("Cantidad de pedidos")

plt.tight_layout()

plt.show()
```

### Interpretación

Los pedidos se distribuyen de forma relativamente equilibrada entre los diferentes estados: Pending, Cancelled, Completed y Refunded.

La presencia de una cantidad importante de pedidos cancelados y reembolsados representa un aspecto relevante para analizar, ya que podría indicar oportunidades de mejora en la gestión de ventas y postventa.

## 5. Distribución del precio total y detección visual de valores atípicos

```
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df["Total Price"]
)

plt.title("Distribución del precio total")
plt.xlabel("Total Price")

plt.tight_layout()

plt.show()
```

### Interpretación

El gráfico de caja permite observar la distribución de los valores correspondientes al precio total de los pedidos e identificar visualmente posibles valores atípicos.

Los valores más elevados representan pedidos de mayor importe. Estos valores fueron analizados junto con el método IQR para determinar si correspondían a errores o a operaciones reales dentro del dataset.

## 5.1 Indicadores principales

### 6.2 Ventas por categoría

```
category_sales = (
    df.groupby("Product Category")["Total Price"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10,5))

sns.barplot(
    data=category_sales,
    x="Total Price",
    y="Product Category"
)

plt.title("Ventas totales por categoría")
plt.xlabel("Ventas totales")
plt.ylabel("Categoría")

plt.tight_layout()
plt.show()
```

### Interpretación

Sports & Outdoors fue la categoría con mayor facturación, seguida por Clothing y Electronics. Beauty & Health registró el menor volumen de ventas. Sin embargo, las diferencias entre las seis categorías fueron moderadas, lo que muestra una distribución relativamente equilibrada de las ventas.

```
status_counts = (
    df["Order Status"]
    .value_counts()
    .reset_index()
)

status_counts.columns = [
    "Order Status",
    "Pedidos"
]

plt.figure(figsize=(8,5))

sns.barplot(
    data=status_counts,
    x="Order Status",
    y="Pedidos"
)

plt.title("Cantidad de pedidos según su estado")
plt.xlabel("Estado del pedido")
plt.ylabel("Cantidad de pedidos")

plt.tight_layout()
plt.show()
```

### Interpretación

Los pedidos se distribuyen de forma relativamente equilibrada entre los cuatro estados disponibles. Pending fue el estado más frecuente, seguido por Cancelled. Completed y Refunded presentaron cantidades ligeramente menores.

```
payment_counts = (
    df["Payment Type"]
    .value_counts()
    .reset_index()
)

payment_counts.columns = [
    "Payment Type",
    "Pedidos"
]

plt.figure(figsize=(8,5))

sns.barplot(
    data=payment_counts,
    x="Payment Type",
    y="Pedidos"
)

plt.title("Cantidad de pedidos por medio de pago")
plt.xlabel("Medio de pago")
plt.ylabel("Cantidad de pedidos")

plt.tight_layout()
plt.show()
```

### Interpretación

PayPal fue el medio de pago más utilizado, con 2.573 pedidos, seguido por Debit Card con 2.529. Credit Card y Gift Card registraron cantidades similares. La distribución relativamente equilibrada indica que los pedidos no se concentran exclusivamente en un único medio de pago.

```
category_status = pd.crosstab(
    df["Product Category"],
    df["Order Status"]
)

category_status.plot(
    kind="bar",
    stacked=True,
    figsize=(11,6)
)

plt.title("Estado de los pedidos por categoría")
plt.xlabel("Categoría")
plt.ylabel("Cantidad de pedidos")

plt.xticks(rotation=45)
plt.legend(title="Estado del pedido")

plt.tight_layout()
plt.show()
```

### Interpretación

La distribución de los estados de pedido es relativamente similar entre las diferentes categorías. Home & Kitchen presenta una proporción ligeramente mayor de pedidos cancelados, mientras que Clothing presenta una mayor proporción de pedidos reembolsados. Electronics presenta una proporción ligeramente superior de pedidos completados. En general, no se observan diferencias extremadamente marcadas entre las categorías.
